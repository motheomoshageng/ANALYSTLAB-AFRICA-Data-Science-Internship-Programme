```python
# Bar chart(2) : Age Distribution

#attrition by department
plt.figure(figsize=(10,6))
sns.countplot(data=df , x='Department',hue='Attrition',palette=['#51cf66', '#ff6b6b'])
plt.title('Employee Attrition by Department',fontsize =14, fontweight='bold')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.show()

#shows which departments are losing people
```

```python
# attrition by Gender BAR GRAPH
plt.figure(figsize=(8,6))
sns.countplot(data=df,x='Gender',hue='Attrition',palette=['#51cf66','#ff6b6b'])
plt.title('Attrition by Gender',fontsize=14, fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('count')
plt.legend(title='Attrition')
plt.tight_layout()
plt.show()
#demographic insights
```

```python
# attrition by JOB ROLE  BAR GRAPH
plt.figure(figsize=(8,6))
df['JobRole'].value_counts().plot(kind='bar',color='#4dabf7')
plt.title('Attrition by JOB ROLES DISTRIBUTION',fontsize=14, fontweight='bold')
plt.xlabel('Job Role')
plt.ylabel('Number of employees')
plt.xticks(rotation = 45,ha='right')
for i, v in enumerate(df['JobRole'].value_counts().values):
    plt.text(i,v +5,str(v),ha='center',fontweight='bold')
plt.tight_layout()
plt.show()
#identifies high risk roles
```

```python
#histogram
#years in cureent role
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(df['Age'], bins=20, edgecolor='black', color='#4dabf7', alpha=0.7)
axes[0].set_title('Age Distribution - All Employees', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Age')
axes[0].set_ylabel('Count')
axes[0].axvline(df['Age'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {df["Age"].mean():.1f}')
axes[0].legend()

sns.histplot(data=df, x='Age', hue='Attrition', bins=20, kde=True, alpha=0.6, palette=['#51cf66', '#ff6b6b'], ax=axes[1])
axes[1].set_title('Age Distribution by Attrition', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Count')
plt.tight_layout()
plt.savefig('age_distribution.png', dpi=300)
plt.show()
```

```python
#Monthly Income Distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(df['MonthlyIncome'], bins=30, edgecolor='black', color='#51cf66', alpha=0.7)
axes[0].set_title('Monthly Income Distribution - All Employees', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Monthly Income ($)')
axes[0].set_ylabel('Count')
axes[0].axvline(df['MonthlyIncome'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: ${df["MonthlyIncome"].mean():,.0f}')
axes[0].legend()

sns.histplot(data=df, x='MonthlyIncome', hue='Attrition', bins=30, kde=True, alpha=0.6, palette=['#51cf66', '#ff6b6b'], ax=axes[1])
axes[1].set_title('Monthly Income by Attrition', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Monthly Income ($)')
axes[1].set_ylabel('Count')
plt.tight_layout()
plt.savefig('income_distribution.png', dpi=300)
plt.show()
```

```python
#box plot - age by attrition
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='Attrition', y='Age')
plt.title('Age Distribution by Attrition Status', fontsize=14, fontweight='bold')
plt.xlabel('Attrition Status')
plt.ylabel('Age (Years)')
plt.show()
```

```python
# Pie chart - Gender
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(gender_counts.values, 
        labels=['Male', 'Female'], 
        autopct='%1.1f%%',
        colors=['#4dabf7', '#f06595'],
        explode=(0, 0.05),
        shadow=True,
        startangle=90)
plt.title('Gender Distribution', fontsize=14, fontweight='bold')
plt.show()
```

```python
# Pie chart - Overtime
overtime_counts = df['OverTime'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(overtime_counts.values, 
        labels=['No Overtime', 'Overtime'], 
        autopct='%1.1f%%',
        colors=['#51cf66', '#ff6b6b'],
        explode=(0, 0.05),
        shadow=True,
        startangle=90)
plt.title('Overtime Distribution', fontsize=14, fontweight='bold')
plt.show()
```
